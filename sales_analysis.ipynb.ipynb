```python
df = df.drop_duplicates()

print("Dataset cleaned!")
print("Dataset size:", df.shape)
```

```python
total_quantity = df['Quantity'].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
total_profit = df['Profit'].sum()

print("Total Profit:", round(total_profit, 2))
```

```python
product_sales = df.groupby('Product Name')['Sales'].sum()

top_products = product_sales.sort_values(
    ascending=False
).head(10)

print(top_products)
```

```python
import matplotlib.pyplot as plt

top_products.sort_values().plot(kind='barh')

plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')

plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

print(region_sales.sort_values(ascending=False))
```

```python
region_sales.sort_values().plot(kind='barh')

plt.title('Sales by Region')
plt.xlabel('Total Sales')
plt.ylabel('Region')

plt.tight_layout()
plt.show()
```

```python
print(df.columns.tolist())
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    errors='coerce'
)

print("Date conversion completed!")
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['Sales'].sum()

print(monthly_sales)
```

```python
import matplotlib.pyplot as plt

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

print(category_sales.sort_values(ascending=False))
```

```python
category_sales.sort_values().plot(kind='bar')

plt.title('Sales by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum()

print(category_profit.sort_values(ascending=False))
```

```python
category_profit.sort_values().plot(kind='bar')

plt.title('Profit by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_sales_df = monthly_sales.reset_index()

monthly_sales_df['Month'] = (
    monthly_sales_df['Month'].astype(str)
)

monthly_sales_df.to_csv(
    '/content/monthly_sales_analysis.csv',
    index=False
)

print("Monthly sales results saved successfully!")
```
